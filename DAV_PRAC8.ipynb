# **Practical-8| Create interactive visualizations using Plotly Express and Plotly Graph Objects.**

**Step-1**: **Import the require libraries**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
```

```python
data = {
    "Month":["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"],
    "Product":["Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet"],
    "Region":["North","South","East","West","North","South","East","West","North","South","East","West"],
    "Sales":[120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Units_Sold":[12,30,25,14,35,30,18,38,32,21,45,40],
    "Profit":[18000,27000,30000,22000,32000,35000,28000,34000,38000,33000,46000,50000]
}

df = pd.DataFrame(data)
df
```

**Step-2: Create an Interactive Line Chart Using Plotly Express**

```python
# Create interactive line chart
fig = px.line(
    df,
    x="Month",
    y="Sales",
    markers=True,
    title="Sales Revenue Throughout the Year"
)

# Add axis labels
fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Sales Revenue (₹)"
)

# Display interactive chart
fig.show()
```

**Step-4: Improve the Hover Information**

```python
# Add additional columns to the hover information.Studentrs can see product, region, Units sold, and profit when they move the mouse over a data point.

# Create interactive line chart
fig = px.line(
    df,
    x="Month",
    y="Sales",
    markers=True,
    title="Monthly Sales Revenue",
    hover_data={
        "Product": True,
        "Region": True,
        "Units_Sold": True,
        "Profit": True,
        "Sales": True
    }
)

# Axis labels
fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Sales Revenue (₹)"
)

# Display chart
fig.show()
```

**Step-5: Create an Interactive Bar Chart**

```python
# Group data by Product and calculate total sales
product_sales = df.groupby("Product", as_index=False)["Sales"].sum()

print(product_sales)

# Create interactive bar chart
fig = px.bar(
    product_sales,
    x="Product",
    y="Sales",
    title="Total Sales by Product",
    text="Sales"
)

# Add axis labels
fig.update_layout(
    xaxis_title="Product",
    yaxis_title="Total Sales (₹)"
)

# Display chart
fig.show()
```

**Step-6: Interactive Scatter Plot**

```python
# To analyze the relationship between unit sold and sales using an intearctive scatter plot

# Create interactive scatter plot
fig = px.scatter(
    df,
    x="Units_Sold",
    y="Sales",
    title="Relationship Between Units Sold and Sales",
    hover_data=["Month", "Product", "Region", "Profit"],
    size="Units_Sold"
)

# Add axis labels
fig.update_layout(
    xaxis_title="Units Sold",
    yaxis_title="Sales Revenue (₹)"
)

# Display chart
fig.show()
```

**Step-7: Interactive Scatter Plot with Color**

```python
# we will use region as the color variable

# Create interactive scatter plot
fig = px.scatter(
    df,
    x="Units_Sold",
    y="Sales",
    color="Region",
    hover_data=["Month", "Product", "Profit"],
    title="Units Sold vs Sales by Region"
)

fig.show()
```

**Step-8:Create an Interactive Pie Chart**


```python
#  To visualize regional sales contibution using an interactive Pie chart.

# Calculate total sales for each region
region_sales = df.groupby("Region", as_index=False)["Sales"].sum()

# Create interactive pie chart
fig = px.pie(
    region_sales,
    names="Region",
    values="Sales",
    title="Regional Sales Contribution",
    hover_data=["Sales"],
    hole=0
)
fig.show()
```

**Step-9: Plotly Graph object**

```python
# To create an interactive visualization using plotly graph objects and understand how it differs from plotly Express.
fig = go.Figure(
    data=[
        go.Scatter(
            x=df['Month'],
            y=df['Sales'],
            mode="lines+markers",
            name="Sales"
        )
    ]
)

fig.update_layout(
    title="Monthly Sales Analysis",
    xaxis_title="Month",
    yaxis_title="Sales"
)

fig.show()
```

**Step-10: Create Line Chart using graph object**

```python
# Create a figure object.
fig = go.Figure()

# Add a line trace to the figure.This trace represents the monthly sales data.
fig.add_trace(
    go.Scatter(
        x=df['Month'],
        y=df['Sales'],
        mode='lines+markers',
        name='Sales'
    )
)

fig.show()
```

**Step 2: Import libraries**

```python
# Import Pandas to handle data
import pandas as pd
# Import Plotly Express to create graphs
import plotly.express as px
# Import Dash components to create the dashboard
from dash import Dash, dcc, html, Input, Output
```

**Step 3: Create the business dataset**

```python
data = {
    "Date": [
        "2026-01-05", "2026-01-10", "2026-01-15",
        "2026-02-03", "2026-02-12", "2026-02-20",
        "2026-03-04", "2026-03-11", "2026-03-25",
        "2026-04-05", "2026-04-14", "2026-04-22",
        "2026-05-02", "2026-05-13", "2026-05-26",
        "2026-06-04", "2026-06-16", "2026-06-25",
        "2026-07-03", "2026-07-15", "2026-07-27",
        "2026-08-05", "2026-08-18", "2026-08-25"
    ],

    "Product": [
        "Rice", "Milk", "Sugar", "Wheat", "Soap", "Oil",
        "Biscuits", "Rice", "Shampoo", "Milk", "Sugar",
        "Wheat", "Oil", "Soap", "Biscuits", "Rice",
        "Milk", "Shampoo", "Sugar", "Oil", "Wheat",
        "Biscuits", "Soap", "Rice"
    ],

    "Category": [
        "Grocery", "Dairy", "Grocery", "Grocery",
        "Personal Care", "Grocery", "Snacks", "Grocery",
        "Personal Care", "Dairy", "Grocery", "Grocery",
        "Grocery", "Personal Care", "Snacks", "Grocery",
        "Dairy", "Personal Care", "Grocery", "Grocery",
        "Grocery", "Snacks", "Personal Care", "Grocery"
    ],

    "Quantity": [
        20, 30, 15, 25, 18, 12, 40, 22,
        10, 35, 20, 28, 15, 25, 45, 30,
        40, 12, 18, 20, 32, 50, 30, 25
    ],

    "Price": [
        60, 30, 45, 40, 35, 150, 10, 60,
        120, 30, 45, 40, 150, 35, 10, 60,
        30, 120, 45, 150, 40, 10, 35, 60
    ],

    "Cost": [
        45, 22, 35, 30, 25, 120, 7, 45,
        90, 22, 35, 30, 120, 25, 7, 45,
        22, 90, 35, 120, 30, 7, 25, 45
    ],

    "Region": [
        "North", "South", "East", "West",
        "North", "South", "East", "West",
        "North", "South", "East", "West",
        "North", "South", "East", "West",
        "North", "South", "East", "West",
        "North", "South", "East", "West"
    ]
}

# Convert dictionary into DataFrame
df = pd.DataFrame(data)

# Convert Date into datetime format
df["Date"] = pd.to_datetime(df["Date"])

# Calculate sales
df["Sales"] = df["Quantity"] * df["Price"]

# Calculate profit
df["Profit"] = df["Quantity"] * (
    df["Price"] - df["Cost"]
)

# Extract month
df["Month"] = df["Date"].dt.strftime("%b")

# Display dataset
df.head(10)
```

**Step 4: Create the dashboard layout**

```python
app = Dash(__name__)

app.title = "General Store Dashboard"

# Design the dashboard layout
app.layout = html.Div([

    # Dashboard Heading
    html.H1(
        "GENERAL STORE BUSINESS DASHBOARD",
        style={
            "textAlign": "center",
            "color": "#17365D"
        }
    ),

    html.P(
        "Sales and Profit Analysis",
        style={"textAlign": "center"}
    ),

    html.Hr(),

    # Category and Region Filters
    html.Div([

        html.Div([
            html.Label("Select Category"),

            dcc.Dropdown(
                id="category",
                options=[
                    {"label": "All Categories", "value": "All"}
                ] + [
                    {"label": c, "value": c}
                    for c in sorted(df["Category"].unique())
                ],
                value="All",
                clearable=False
            )

        ], style={"width": "45%"}),

        html.Div([
            html.Label("Select Region"),

            dcc.Dropdown(
                id="region",
                options=[
                    {"label": "All Regions", "value": "All"}
                ] + [
                    {"label": r, "value": r}
                    for r in sorted(df["Region"].unique())
                ],
                value="All",
                clearable=False
            )

        ], style={"width": "45%"})

    ], style={
        "display": "flex",
        "justifyContent": "space-around",
        "margin": "20px"
    }),

    # KPI Cards
    html.Div(
        id="kpi-cards",
        style={
            "display": "flex",
            "justifyContent": "space-around",
            "gap": "15px",
            "flexWrap": "wrap"
        }
    ),

    html.Br(),

    # First Row of Charts
    html.Div([

        html.Div([
            dcc.Graph(id="sales-chart")
        ], style={"width": "48%"}),

        html.Div([
            dcc.Graph(id="category-chart")
        ], style={"width": "48%"})

    ], style={
        "display": "flex",
        "flexWrap": "wrap",
        "justifyContent": "space-around"
    }),

    # Second Row of Charts
    html.Div([

        html.Div([
            dcc.Graph(id="profit-chart")
        ], style={"width": "48%"}),

        html.Div([
            dcc.Graph(id="region-chart")
        ], style={"width": "48%"})

    ], style={
        "display": "flex",
        "flexWrap": "wrap",
        "justifyContent": "space-around"
    }),

    # Sales Records
    html.H2("General Store Sales Records"),

    html.Div(id="data-table")

], style={
    "backgroundColor": "#F4F6F9",
    "padding": "25px",
    "fontFamily": "Arial"
})
```

**Step 5: Add interactive dashboard functionality**

```python
# ---------------------------------------
# KPI CARD FUNCTION
# ---------------------------------------

def create_card(title, value):

    return html.Div([

        html.H4(title, style={"color": "#555"}),

        html.H2(value, style={"color": "#1677B8"})

    ], style={
        "backgroundColor": "white",
        "padding": "15px",
        "width": "20%",
        "minWidth": "150px",
        "textAlign": "center",
        "borderRadius": "10px",
        "boxShadow": "0 2px 6px rgba(0,0,0,0.1)"
    })


# ---------------------------------------
# INTERACTIVE DASHBOARD CALLBACK
# ---------------------------------------

@app.callback(

    Output("kpi-cards", "children"),
    Output("sales-chart", "figure"),
    Output("category-chart", "figure"),
    Output("profit-chart", "figure"),
    Output("region-chart", "figure"),
    Output("data-table", "children"),

    Input("category", "value"),
    Input("region", "value")

)

def update_dashboard(selected_category, selected_region):

    # Filter dataset
    filtered = df.copy()

    if selected_category != "All":
        filtered = filtered[
            filtered["Category"] == selected_category
        ]

    if selected_region != "All":
        filtered = filtered[
            filtered["Region"] == selected_region
        ]

    # -----------------------------------
    # CALCULATE KPIs
    # -----------------------------------

    total_sales = filtered["Sales"].sum()
    total_profit = filtered["Profit"].sum()
    total_quantity = filtered["Quantity"].sum()
    total_products = filtered["Product"].nunique()

    kpis = [

        create_card("Total Sales", f"₹{total_sales:,.0f}"),

        create_card("Total Profit", f"₹{total_profit:,.0f}"),

        create_card("Items Sold", f"{total_quantity:,}"),

        create_card("Products", str(total_products))

    ]

    # -----------------------------------
    # MONTHLY SALES CHART
    # -----------------------------------

    monthly = filtered.groupby(
        "Month", as_index=False
    )["Sales"].sum()

    month_order = [
        "Jan", "Feb", "Mar", "Apr",
        "May", "Jun", "Jul", "Aug",
        "Sep", "Oct", "Nov", "Dec"
    ]

    monthly["Month"] = pd.Categorical(
        monthly["Month"],
        categories=month_order,
        ordered=True
    )

    monthly = monthly.sort_values("Month")

    sales_fig = px.line(
        monthly,
        x="Month",
        y="Sales",
        markers=True,
        title="Monthly Sales Analysis",
        template="plotly_white"
    )

    # -----------------------------------
    # CATEGORY-WISE SALES CHART
    # -----------------------------------

    category_data = filtered.groupby(
        "Category", as_index=False
    )["Sales"].sum()

    category_fig = px.bar(
        category_data,
        x="Category",
        y="Sales",
        color="Category",
        title="Sales by Category",
        template="plotly_white"
    )

    # -----------------------------------
    # PRODUCT-WISE PROFIT CHART
    # -----------------------------------

    product_data = filtered.groupby(
        "Product", as_index=False
    )["Profit"].sum()

    profit_fig = px.bar(
        product_data,
        x="Product",
        y="Profit",
        color="Profit",
        title="Product-wise Profit",
        template="plotly_white"
    )

    # -----------------------------------
    # REGION-WISE SALES CHART
    # -----------------------------------

    region_data = filtered.groupby(
        "Region", as_index=False
    )["Sales"].sum()

    region_fig = px.pie(
        region_data,
        names="Region",
        values="Sales",
        title="Region-wise Sales",
        hole=0.4
    )

    # -----------------------------------
    # DISPLAY SALES DATA TABLE
    # -----------------------------------

    table = html.Table([

        html.Thead(
            html.Tr([
                html.Th(col)
                for col in [
                    "Date", "Product", "Category",
                    "Quantity", "Sales", "Profit", "Region"
                ]
            ])
        ),

        html.Tbody([

            html.Tr([

                html.Td(str(row.Date.date())),
                html.Td(row.Product),
                html.Td(row.Category),
                html.Td(row.Quantity),
                html.Td(f"₹{row.Sales:,.0f}"),
                html.Td(f"₹{row.Profit:,.0f}"),
                html.Td(row.Region)

            ])

            for row in filtered.itertuples()

        ])

    ], style={
        "width": "100%",
        "backgroundColor": "white",
        "borderCollapse": "collapse",
        "textAlign": "center"
    })

    # Return all dashboard components

    return (
        kpis,
        sales_fig,
        category_fig,
        profit_fig,
        region_fig,
        table
    )
```

# **Practical-6| Create basic data visualizations (Line, Bar, Histogram, Scatter, Pie).**

**Step-1: Import require library**

```python
# Import Pandas for data manipulation and analysis
import pandas as pd
# Import NumPy for numerical calculations
import numpy as np
# Import Matplotlib for creating graphs and visualizations
import matplotlib.pyplot as plt
```

**Step-2: Create a Dataset**

```python
data = {
    "Month":["jan","feb","mar","apr","may","jun","jul","aug","sep","oct","nov","dec"],
    "Product":["Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet"],
    "Region":["North","South","East","West","North","South","East","West","North","South","East","West"],
    "Sales":[120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Units_Sold":[12,30,25,14,35,30,18,38,32,21,45,40]
}

df = pd.DataFrame(data)
df
```

**Step-3: Create a line chart**

**Task: How did the company sales changes from jan to dec**

```python
plt.plot(df['Month'], df['Sales'], marker='o', color='blue', linewidth=2)

# Display sales values
for month, value in zip(df['Month'], df['Sales']):
    plt.text(month, value + 5000, f"₹{value:,}", ha='center')

plt.title("Company Sales from January to December")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

**Step-4: Create a bar chart**

**Task: Which product generates how much total sales.**

```python
# Calculate total sales for each product
total_sales = df.groupby("Product")["Sales"].sum()

# Create bar chart
plt.figure(figsize=(8, 5))
total_sales.plot(kind="bar")

plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.show()
```

**Step-5: Create a Histogram**

**Task: How are the sales values individual distributed**

```python
# Set the size of the figure
plt.figure(figsize=(8, 5))

# Create a histogram using the Sales column
# bins=5 divides the sales values into 5 ranges
# edgecolor="black" adds black borders around the bars
plt.hist(df["Sales"], bins=5, edgecolor="black")

# Add a title to the histogram
plt.title("Distribution of Sales Values")

# Label the X-axis
plt.xlabel("Sales")

# Label the Y-axis
# Frequency means how many sales values fall into each range
plt.ylabel("Frequency")

# Display the histogram
plt.show()
```

**Step-5: Create a scatter plot**

**Task: Is their a relationship between the no. of unit sales and total sales.**

```python
# Set the chart size
plt.figure(figsize=(8, 5))

# Create scatter plot
plt.scatter(df["Units_Sold"], df["Sales"])

# Add chart title
plt.title("Relationship Between Units Sold and Sales")

# Label X-axis
plt.xlabel("Units Sold")

# Label Y-axis
plt.ylabel("Sales")

# Display grid lines
plt.grid(True)

# Display the chart
plt.show()
```

**Step-6: Create a pie chart**

**Task: What percentage of total sales comes from each region**

```python
# Calculate total sales for each region
region_sales = df.groupby("Region")["Sales"].sum()

# Create pie chart
plt.figure(figsize=(7, 7))

plt.pie(
    region_sales,
    labels=region_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Percentage of Total Sales by Region")

plt.show()
```

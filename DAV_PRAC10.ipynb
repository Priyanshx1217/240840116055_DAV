**Step 1 : Install and import libraries**

```python
# Install required libraries
!pip install -q pandas plotly dash

# Import pandas for data handling
import pandas as pd

# Import plotly express for visualizations
import plotly.express as px

# Import Dash components for dashboard development
from dash import Dash, dcc, html, Input, Output
```

**Step 2 : Load the real world Dataset**

```python
# URL of the public Sample Superstore dataset
url = "https://raw.githubusercontent.com/johnreygoh/datasets/master/SampleSuperstore.csv"

# Load the CSV dataset into a DataFrame
original_df = pd.read_csv(url)

# Select 2000 random records
df = original_df.sample(n=2000, random_state=42)

# Reset the index
df = df.reset_index(drop=True)

# Display the first five records
print("First Five Records:")
display(df.head())

# Display dataset dimensions
print("\nDataset Shape:", df.shape)
```

**Step 3 : Understand the dataset**

```python
# Display all column names
print("Column Names:")
print(df.columns.tolist())

# Display dataset information
print("\nDataset Information:")
df.info()

# Display statistical summary
print("\nStatistical Summary:")
display(df.describe())
```

**Step 4 : Data cleaning and preprocessing**

```python
# Check missing values in each column
print("Missing Values:")
print(df.isnull().sum())

# Check duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())

# Remove duplicate records
df = df.drop_duplicates()

# Remove rows with missing important values
df = df.dropna(subset=['Sales', 'Profit', 'Category', 'Region'])

# Convert sales, profit and quantity into numeric format
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
df['Profit'] = pd.to_numeric(df['Profit'], errors='coerce')
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')

# Remove records with invalid numeric values
df = df.dropna(subset=['Sales', 'Profit', 'Quantity'])

# Display cleaned dataset
print("\nCleaned Dataset:")
display(df.head())

# Display final number of records
print("\nFinal Dataset Shape:", df.shape)
```

**Step 5 : Exploratory Data Analysis**

```python
# Calculate total sales
total_sales = df['Sales'].sum()

# Calculate total profit
total_profit = df['Profit'].sum()

# Calculate total quantity sold
total_quantity = df['Quantity'].sum()

# Calculate average sales
average_sales = df['Sales'].mean()

# Display business statistics
print("TOTAL SALES:", round(total_sales, 2))
print("TOTAL PROFIT:", round(total_profit, 2))
print("TOTAL QUANTITY SOLD:", total_quantity)
print("AVERAGE SALES:", round(average_sales, 2))

# Calculate sales by category
print("\nSales by Category:")
display(df.groupby('Category')['Sales'].sum())

# Calculate sales by region
print("\nSales by Region:")
display(df.groupby('Region')['Sales'].sum())

# Calculate profit by sub-category
print("\nProfit by Sub-Category:")
display(df.groupby('Sub-Category')['Profit'].sum())
```

**Step 6 : Create data visualizations**

```python
# Calculate total sales for each category
category_sales = df.groupby(
    'Category', as_index=False
)['Sales'].sum()

# Create an interactive bar chart
fig1 = px.bar(
    category_sales,
    x='Category',
    y='Sales',
    color='Category',
    title='Sales by Category',
    text_auto='.2s'
)

# Adjust chart layout
fig1.update_layout(
    yaxis_title="Sales (₹)",
    template='plotly_white',
    height=400
)

# Display the chart
fig1.show()

# Calculate total sales for each region
region_sales = df.groupby(
    'Region', as_index=False
)['Sales'].sum()

# Create an interactive pie chart
fig2 = px.pie(
    region_sales,
    names='Region',
    values='Sales',
    title='Sales Distribution by Region',
    hole=0.35
)

# Display profit values in rupees on hover
fig2.update_traces(
    hovertemplate='%{label}<br>Profit: ₹%{value:,.2f}<extra></extra>'
)

# Display the chart
fig2.show()

# Calculate total profit for each sub-category
subcategory_profit = df.groupby(
    'Sub-Category', as_index=False
)['Profit'].sum()

# Sort sub-categories according to profit
subcategory_profit = subcategory_profit.sort_values(
    by='Profit', ascending=False
)

# Create an interactive horizontal bar chart
fig3 = px.bar(
    subcategory_profit,
    x='Profit',
    y='Sub-Category',
    orientation='h',
    color='Profit',
    title='Profit by Sub-Category'
)

# Display profit values in rupees
fig3.update_layout(
    xaxis_title="Profit (₹)",
    template="plotly_white"
)

# Display the chart
fig3.show()
```

**Step 7 : Design the Dashboard Layout**

```python
# Create the Dash application
app = Dash(__name__)

# Define the dashboard layout
app.layout = html.Div([

    # Main dashboard heading
    html.H1(
        "SUPERSTORE SALES DASHBOARD",
        style={
            'textAlign': 'center',
            'color': 'white'
        }
    ),

    # Dashboard subtitle
    html.H3(
        "Retail Business Performance Analysis",
        style={
            'textAlign': 'center',
            'color': '#00D4FF'
        }
    ),
    html.Br(),

    # Region filter heading
    html.Label(
        "Select Region",
        style={'color': 'white'}
    ),

    # Region dropdown
    dcc.Dropdown(
        id='region-filter',
        options=[
            {'label': 'All Regions', 'value': 'All'}
        ] + [
            {'label': r, 'value': r}
            for r in sorted(df['Region'].dropna().unique())
        ],
        value='All',
        clearable=False
    ),
    html.Br(),

    # Category filter heading
    html.Label(
        "Select Category",
        style={'color': 'white'}
    ),

    # Category dropdown
    dcc.Dropdown(
        id='category-filter',
        options=[
            {'label': 'All Categories', 'value': 'All'}
        ] + [
            {'label': c, 'value': c}
            for c in sorted(df['Category'].dropna().unique())
        ],
        value='All',
        clearable=False
    ),
    html.Br(),

    # KPI cards section
    html.Div([

        # Total sales card
        html.Div([
            html.H4("Total Sales"),
            html.H2(id='sales-kpi')
        ], style={
            'backgroundColor': '#243B55',
            'color': 'white',
            'padding': '15px',
            'width': '30%',
            'textAlign': 'center',
            'borderRadius': '10px'
        }),

        # Total profit card
        html.Div([
            html.H4("Total Profit"),
            html.H2(id='profit-kpi')
        ], style={
            'backgroundColor': '#243B55',
            'color': 'white',
            'padding': '15px',
            'width': '30%',
            'textAlign': 'center',
            'borderRadius': '10px'
        }),

        # Total quantity card
        html.Div([
            html.H4("Total Quantity"),
            html.H2(id='quantity-kpi')
        ], style={
            'backgroundColor': '#243B55',
            'color': 'white',
            'padding': '15px',
            'width': '30%',
            'textAlign': 'center',
            'borderRadius': '10px'
        })

    ], style={
        'display': 'flex',
        'justifyContent': 'space-around',
        'gap': '15px'
    }),
    html.Br(),

    # Sales by category chart
    dcc.Graph(id='category-chart'),

    # Sales by region chart
    dcc.Graph(id='region-chart'),

    # Profit by sub-category chart
    dcc.Graph(id='profit-chart')

], style={
    'backgroundColor': '#141E30',
    'padding': '20px'
})
```

**Step 8 : Add Interactivity**

```python
# Define callback to update dashboard components
@app.callback(

    # Output components to be updated
    Output('sales-kpi', 'children'),
    Output('profit-kpi', 'children'),
    Output('quantity-kpi', 'children'),
    Output('category-chart', 'figure'),
    Output('region-chart', 'figure'),
    Output('profit-chart', 'figure'),

    # Input values from dropdown filters
    Input('region-filter', 'value'),
    Input('category-filter', 'value')
)

# Function to update dashboard
def update_dashboard(region, category):

    # Create a copy of the original dataset
    filtered = df.copy()

    # Apply region filter
    if region != 'All':
        filtered = filtered[
            filtered['Region'] == region
        ]

    # Apply category filter
    if category != 'All':
        filtered = filtered[
            filtered['Category'] == category
        ]

    # Calculate KPI values
    total_sales = filtered['Sales'].sum()
    total_profit = filtered['Profit'].sum()
    total_quantity = filtered['Quantity'].sum()

    # Group sales according to category
    c = filtered.groupby(
        'Category', as_index=False
    )['Sales'].sum()

    # Create category sales chart
    fig1 = px.bar(
        c,
        x='Category',
        y='Sales',
        color='Category',
        title='Sales by Category'
    )

    # Group sales according to region
    r = filtered.groupby(
        'Region', as_index=False
    )['Sales'].sum()

    # Create regional sales pie chart
    fig2 = px.pie(
        r,
        names='Region',
        values='Sales',
        title='Sales by Region'
    )

    # Group profit according to sub-category
    s = filtered.groupby(
        'Sub-Category', as_index=False
    )['Profit'].sum()

    # Create sub-category profit chart
    fig3 = px.bar(
        s,
        x='Profit',
        y='Sub-Category',
        orientation='h',
        color='Profit',
        title='Profit by Sub-Category'
    )

    # Return KPI values
    return (
    f"₹{total_sales:,.2f}",
    f"₹{total_profit:,.2f}",
    total_quantity,
    fig1,
    fig2,
    fig3
)
```

**Step 9 : Run the Interactive Dashboard**

```python
# Run the dashboard inside Google Colab
app.run(jupyter_mode="inline", debug=False)
```
